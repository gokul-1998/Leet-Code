```python
class Solution:
    def minAddToMakeValid(self, s: str) -> int:
        stack=[]
        res=0
        for i in s:
            if i =='(':
                stack.append('(')
            else:
                if not stack: # this is the case where we have more ')' than '('
                    res+=1
                else:
                    stack.pop()
        return res+len(stack)
```

# The story is like this, 
-  we have a stack to keep track of the opening brackets. 
- When we encounter a closing bracket, we check if there is a corresponding opening bracket in the stack. 
    - If there is, we pop it from the stack.
    -  If not, we increment the result counter because we need to add an opening bracket to make it valid.
    -  At the end, we add the length of the stack (which contains unmatched opening brackets) to the result counter to get the total number of brackets needed to make the string valid.

```python
# The story is like this, 
# -  we have a stack to keep track of the opening brackets. 
# - When we encounter a closing bracket, we check if there is a corresponding opening bracket in the stack. 
#     - If there is, we pop it from the stack.
#     -  If not, we increment the result counter because we need to add an opening bracket to make it valid.
#     -  At the end, we add the length of the stack (which contains unmatched opening brackets) to the result counter to get the total number of brackets needed to make the string valid.

class Solution:
    def minAddToMakeValid(self, s: str) -> int:
        stack=[]
        res=0
        for i in s:
            if i=="(":
                stack.append('(')
            else:
                if not stack:
                    res+=1
                else:
                    stack.pop()
        return res+len(stack)
```
