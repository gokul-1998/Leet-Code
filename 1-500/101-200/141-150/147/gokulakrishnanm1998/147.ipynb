### 147. Insertion Sort List

- https://leetcode.com/problems/insertion-sort-list/description/

- Given the `head of a singly linked list`, `sort the list using insertion sort`, and `return the sorted list's head`.

- The steps of the insertion sort algorithm:

    - Insertion sort iterates, consuming one input element each repetition and growing a sorted output list.
    - At each iteration, insertion sort removes one element from the input data, finds the location it belongs within the sorted list and inserts it there.
    - It repeats until no input elements remain.
- The following is a graphical example of the insertion sort algorithm. The partially sorted list (black) initially contains only the first element in the list. One element (red) is removed from the input data and inserted in-place into the sorted list with each iteration.



```python
# Create a dummy node

# curr = head

# while curr is not None:

#     Save the next node

#     Start searching from dummy

#     Find the correct insertion position

#     Insert curr into the sorted list

#     Move curr to the saved next node


# Return dummy.next
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
    def __repr__(self):
        return f"ListNode({self.val}|{self.next.val if self.next else None})"

# Why is important and try to connect the dots.
class Solution:
    def insertionSortList(self, head):
        dummy = ListNode(0) 
        # we return the dummy.next so we need a dummy node to always refer to starting point.
        curr = head
        # set starting from the head as current node.

        while curr: # as long as current node exists
            next_node = curr.next # store the next value of current in a next_node variable.as we will be changing the value of curr.next down.

            prev = dummy # prev=dummy

            # Find the correct insertion position
            while prev.next and prev.next.val < curr.val: # while prev.next and its value<curr.val keep moving the prev to its next value
                prev = prev.next

            # Insert current node 
            curr.next = prev.next
            prev.next = curr

            # Move to the next original node
            curr = next_node

        return dummy.next

sol=Solution()
# Example usage: # 4 2 1 3
head = ListNode(4, ListNode(2, ListNode(1, ListNode(3))))
sorted_head = sol.insertionSortList(head)
# Print the sorted linked list
curr = sorted_head
while curr:
    print(curr.val, end=" -> ")
    curr = curr.next
print("None")   
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def insertionSortList(self, head: ListNode | None) -> ListNode | None:
        dummy=ListNode(-5000)
        curr=head
        while curr:
            next_node=curr.next

            prev=dummy
            while prev.next and prev.next.val<=curr.val:
                prev=prev.next
            # prev is the node after which curr is supposed to be inserted.
            curr.next=prev.next # because after inserting, the curren.next needs to be prev.next
            prev.next=curr # then prev.next=curr.

            curr=next_node
        return dummy.next
```
