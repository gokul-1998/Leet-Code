````python
class Solution:
    def longestValidParentheses(self, s: str) -> int:
        stack = [-1]
        highest = 0

        for i, ch in enumerate(s):
            if ch == '(':
                stack.append(i)
            else:
                stack.pop()

                if not stack:
                    stack.append(i)
                else:
                    highest = max(highest, i - stack[-1])

        return highest

# the approach is 
    # - Use a stack to keep track of the indices of the characters in the string.
    # - For each character in the string, if it's an opening parenthesis '(', push its index onto the stack.
    # - If it's a closing parenthesis ')', pop an element from the stack.
    #     - If the stack is empty after popping, it means this closing parenthesis doesn't have a matching opening one, so we push its index onto the stack to mark the start of a new valid substring.
    #     - Otherwise, we calculate the length of the current valid substring by subtracting the index of the last element in the stack from the current index, and update the maximum length if necessary.
````

Your approach is close, but the main problem is that counting matched pairs (`curr += 2`) doesn't guarantee that the parentheses form one continuous valid substring.

For `"()(()"`:

- `"()"` is valid, length `2`.
- The remaining `"(("` has unmatched opening parentheses.
- Your code counts the first pair correctly, but it cannot properly handle valid substrings that appear after unmatched parentheses or combine across earlier matches.

## 1. Recommended approach: Stack of indices

Instead of storing `'('`, store its index. Use `-1` as the initial boundary.

```python
class Solution:
    def longestValidParentheses(self, s: str) -> int:
        stack = [-1]
        highest = 0

        for i, ch in enumerate(s):
            if ch == '(':
                stack.append(i)
            else:
                stack.pop()

                if not stack:
                    stack.append(i)
                else:
                    highest = max(highest, i - stack[-1])

        return highest
```

## 2. How does it work?

Let's trace the input `s = "()(()"`.

| Index | Character | Stack after processing | Maximum length |
| ----- | --------- | ---------------------- | -------------- |
| Start | —         | `[-1]`                 | 0              |
| 0     | `(`       | `[-1, 0]`              | 0              |
| 1     | `)`       | `[-1]`                 | 2              |
| 2     | `(`       | `[-1, 2]`              | 2              |
| 3     | `(`       | `[-1, 2, 3]`           | 2              |
| 4     | `)`       | `[-1, 2]`              | 2              |

At index `1` (the index wherever we see ')'), the valid substring length is:

\\[ 1-(-1)=2 \\]

At index `4` (the index wherever we see '), the valid substring length is:

\\[ 4-2=2 \\]

The answer is `2`.

## 3. Understand the three important operations

- Opening `(`: Push its index onto the stack.
- Closing `)`: Pop the last index because it represents a matched opening parenthesis.
- Empty stack after popping: The current closing parenthesis is an invalid boundary, so push its index as the new boundary.
- Otherwise: Calculate the valid substring length using `i - stack[-1]`.

Key takeaway: For Longest Valid Parentheses, remember this pattern:

`stack of indices + boundary index (-1) + current index - stack top`

Your original approach counts pairs, but this approach tracks the length of a continuous valid substring. It runs in \\(O(n)\\) time and uses \\(O(n)\\) extra space.

````python
# the approach is 
    # - Use a stack to keep track of the indices of the characters in the string.
    # - For each character in the string, if it's an opening parenthesis '(', push its index onto the stack.
    # - If it's a closing parenthesis ')', pop an element from the stack.
    #     - If the stack is empty after popping, it means this closing parenthesis doesn't have a matching opening one, so we push its index onto the stack to mark the start of a new valid substring.
    #     - Otherwise, we calculate the length of the current valid substring by subtracting the index of the last element in the stack from the current index, and update the maximum length if necessary.

class Solution:
    def longestValidParentheses(self, s: str) -> int:
        stack=[-1]
        highest=0
        
        for i,ch in enumerate(s):
            if ch=='(':
                stack.append(i)
            else:
                stack.pop()
                if not stack:
                    stack.append(i)
                else:
                    highest=max(highest,i-stack[-1])
        return highest
                
        
       
                    

````
